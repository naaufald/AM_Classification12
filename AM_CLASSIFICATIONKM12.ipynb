```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score


saham = pd.read_excel("D:/Matana/Semester 5/ANALISIS MULTIVARIAT/saham saham/rangkuman saham.xlsx")
saham
```

```python
data_cluster = saham.iloc[:, 1:]
data_cluster.columns = data_cluster.columns.astype(str)
kluster2 = KMeans(n_clusters=2, random_state=42)
label2 = kluster2.fit_predict(data_cluster)
```

```python
k3 = KMeans(n_clusters=3, random_state=42)
label3 = k3.fit_predict(data_cluster)
k3
```

```python
silhouette_2 = silhouette_score(data_cluster, label2)
print("Silhouette Score (2 cluster):", silhouette_2)
```

```python
silhouette_3 = silhouette_score(data_cluster, label3)
print("Silhouette Score (3 cluster):", silhouette_3)
```

```python
import numpy as np
import skfuzzy as fuzz

# convert ke numpy array
data_np = np.array(data_cluster).T   # Transpose wajib untuk skfuzzy

# -------- 2 Cluster Fuzzy --------
cntr2, u2, u0, d2, jm2, p2, fpc2 = fuzz.cluster.cmeans(
    data_np, c=2, m=2, error=0.005, maxiter=1000, init=None
)

# label fuzzy = cluster dengan membership terbesar
fuzzy_labels_2 = np.argmax(u2, axis=0)

saham['fuzzy2'] = fuzzy_labels_2
print("Fuzzy Partition Coefficient (FPC) 2 cluster:", fpc2)

# -------- 3 Cluster Fuzzy --------
cntr3, u3, u0, d3, jm3, p3, fpc3 = fuzz.cluster.cmeans(
    data_np, c=3, m=2, error=0.005, maxiter=1000, init=None
)

fuzzy_labels_3 = np.argmax(u3, axis=0)

saham['fuzzy3'] = fuzzy_labels_3
print("Fuzzy Partition Coefficient (FPC) 3 cluster:", fpc3)
```

```python
from sklearn.cluster import KMeans
kmeans3 = KMeans(n_clusters=3, random_state=42)
kmeans3.fit(data_cluster)
labels3 = kmeans3.labels_
print("Label Cluster (3 cluster):")
print(labels3)
```
